```python
#there are various methods opf creating DataFrames
 #from a dictionary: 
import pandas as pd
data = {
    "Name": ["Ameeroo", "Raees", "Riaz"],
    "Age" : [20, 21, 51], 
    "Height" : [172, 177, 175]
 }
df = pd.DataFrame(data)
print(df) 

    
```

```python
#create a dataframe from a list
df = pd.DataFrame([
    ["Ameeroo", 20, 172],
                  ["Raees", 21, 177],
                  ["Riaz", 51, 175]], 
                  columns = ["Name", "Age", "Height"])
print(df)
```

```python
#we have created a dataframe above  
#we are learning how to access columns 
#one col : 
print(df["Name"])
print(df[["Age", "Height"]])
print(df[["Name", "Age", "Height"]])
```

```python
#now we access by rows using iloc sand loc
#thinkl of i loc -> index location 

print(df.iloc[0])

#Think opf loc as label
print(df.loc[1])
```

```python
#Now we learn to filter 
#we want rows where age > 30 : 
print(df[df["Age"] >30 ]) 
```

```python
#now we learn modify a Dataframe
# we want a new column
df["Salary"] = [10000, 30000, 100000]
print(df.head())
```

```python
print(df[df["Salary"] == 100000])
```

```python
#Now we are going to update values  
#we first go to the location that we are updating and we update
df.loc[0, "Age"] = 21
print(df[df["Age"] == 21]) 
```

```python
#deleting columns 
df["Height"] = [172, 177, 175]
print(df.head())

df.drop(columns = ["Height"], inplace = True)
print(df.head())
```

```python
#Lets learn some commin dataframe operations 
#basic statistics 
print(df.describe())

print(df.info())
```

```python
#sorting
df_sorted = df.sort_values(by = "Age", ascending = False)
print(df_sorted)

#grouping
df_grouped = df.groupby("Name").mean()
print(df_grouped) 
```
